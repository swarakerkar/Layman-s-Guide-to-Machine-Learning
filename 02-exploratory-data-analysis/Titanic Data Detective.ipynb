## Survival Analysis

```python
survival_counts = titanic["Survived"].value_counts()

print("Survival counts:")
print(survival_counts)
```

```python
print("Survival counts with labels:")

print("Did not survive:", survival_counts.get(0, 0))
print("Survived:", survival_counts.get(1, 0))
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=titanic, x="Survived", hue="Survived", palette="Set2", legend=False)

plt.title("Number of Survivors and Non-Survivors")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.xticks([0, 1], ["Did Not Survive", "Survived"])

plt.show()
```

## Analysis of Passenger Class

```python
print("Passenger class counts:")
print(titanic["Pclass"].value_counts().sort_index())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=titanic,
    x="Pclass",
    hue="Pclass",
    palette="viridis",
    legend=False
)

plt.title("Number of Passengers in Each Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

## Gender Analysis

```python
gender_counts = titanic["Sex"].value_counts()

print("Gender counts:")
print(gender_counts)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=titanic,
    x="Sex",
    hue="Sex",
    palette="Set2",
    legend=False
)

plt.title("Distribution of Passengers by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

## Embarkation Town Analysis

```python
number_of_towns = titanic["Embarked"].nunique()

print("Number of unique embarkation towns:", number_of_towns)
```

```python
print("Embarkation town frequencies:")
print(titanic["Embarked"].value_counts())
```

## Missing Data Analysis

```python
missing_values = titanic.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
print("Percentage of missing values in each column:")

missing_percentage = (titanic.isnull().sum() / len(titanic)) * 100

display(
    missing_percentage
    .sort_values(ascending=False)
    .to_frame("Missing Percentage")
)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(
    titanic.isnull(),
    yticklabels=False,
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Values in the Titanic Dataset")
plt.xlabel("Features")
plt.ylabel("Passengers")

plt.show()
```

## Box plot for age and outlier detection

## Correlation Analysis

```python
# Select only numerical columns
numerical_data = titanic.select_dtypes(include=np.number)

print("Numerical columns used for correlation analysis:")
print(numerical_data.columns.tolist())
```

```python
correlation = numerical_data.corr()

print("Correlation matrix:")
display(correlation)
```

```python
print("Correlation of numerical features with survival:")
display(correlation["Survived"].sort_values(ascending=False))
```

## Heatmap

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="crest",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")

plt.show()
```

## Numerical Data Distribution Analysis

```python
numerical_columns = ["Age", "Fare", "SibSp", "Parch"]

titanic[numerical_columns].hist(
    figsize=(12, 8),
    bins=20,
    color="steelblue",
    edgecolor="black"
)

plt.suptitle("Distribution of Numerical Features", fontsize=16)

plt.tight_layout()

plt.show()
```

## Gender-wise Survival Rate

```python
survival_by_gender = titanic.groupby("Sex", observed=True)["Survived"].mean()

print("Survival rate by gender:")
display(survival_by_gender)
```

```python
plt.figure(figsize=(7, 5))

survival_by_gender.plot(
    kind="bar",
    color=["steelblue", "lightcoral"]
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.xticks(rotation=0)

plt.ylim(0, 1)

plt.show()
```
